```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import confusion_matrix
from sklearn.metrics import f1_score
from sklearn.metrics import accuracy_score
```

```python
dataset = pd.read_csv('diabetes.csv')
#dataset = dataset[~(dataset['Insulin'] == 0)]
print(len(dataset))
dataset.head()
#Mang thai    #Đường Glucose  #Huyết áp #độ dày của da #Insulin #Chỉ số khối cơ thể(BMI)
#DiabetesPedigreeFunction: một chức năng cho điểm khả năng mắc bệnh tiểu đường dựa trên tiền sử gia đình)
```

```python
# Thay thế số 0 trong các cột
zero_not_accepted = ['Glucose', 'BloodPressure', 'SkinThickness', 'BMI', 'Insulin']
```

```python
for column in zero_not_accepted:
    dataset[column] = dataset[column].replace(0, np.NaN)
    mean = int(dataset[column].mean(skipna=True))
    dataset[column] = dataset[column].replace(np.NaN, mean)
```

```python
# chia dữ liệu train và test
X = dataset.iloc[:, :-1]
y = dataset.iloc[:, -1]
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0, test_size=0.2)#20% dữ liệu để test
```
