```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def gradient_function(x):
    return 2*x
    
def gradient_descent(gradient, start, learning_rate = 0.01, iterations = 1000, stopping_threshold = 1e-6):
    data = []
    data_ = start
    data.append(data_)
    for _ in range(iterations):
        diff = -learning_rate * gradient(data_)
        if np.all(np.abs(diff) <= stopping_threshold):
            break
        data_ += diff
        data.append(data_)
    
X = np.arange(-10, 10, 0.001)
plt.plot(X, np.sin(X))
#plt.plot(data, np.square(data), '-o', color='red', linewidth='0.5')
plt.ylabel("F(x)")
plt.xlabel("x")
plt.show()
#print(len(data))
```

```python
gradient_descent(gradient=gradient_function, start=-5, learning_rate=0.4)
```
