```python
import math
import numpy as np
import matplotlib.pyplot as plt
def gradient_sin(x):
  return math.cos(x)


def gradient_descent_sin(gradient, start, learning_rate = 0.01, iterations = 1000, stopping_threshold = 1e-6):
    data = []
    data_ = start
    data.append(data_)
    for _ in range(iterations):
        diff = -learning_rate * gradient_sin(data_)
        if np.all(np.abs(diff) <= stopping_threshold):
            break
        data_ += diff
        data.append(data_)

    print(len(data))

    x = np.arange(0, 5*np.pi, 0.1)
    y = np.sin(x)
    plt.plot(x, y)
    plt.plot(data, np.sin(data), '-o', color='red', linewidth='0.5')
    plt.ylabel("F(x)")
    plt.xlabel("x")
    plt.show()

i = 0.2
while i < 1:
  gradient_descent_sin(gradient=gradient_sin, start=2, learning_rate=i)
  i+=0.2
```
