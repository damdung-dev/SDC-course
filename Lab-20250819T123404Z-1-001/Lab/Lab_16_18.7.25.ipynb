```python
import numpy as np
import matplotlib.pyplot as plt
import math
%matplotlib inline
```

```python
# Draw function
def draw(x, y, x_lim, y_lim, multi = False, label = None):
    fig, ax = plt.subplots()

    # set x and y limits
    xlim = [-x_lim, x_lim]
    ylim = [-y_lim, y_lim]
    ax.set_xlim(xlim)
    ax.set_ylim(ylim)

    # draw x and y axes
    ax.axhline(y=0, color='black')
    ax.axvline(x=0, color='black')

    # add labels to axes
    ax.set_xlabel('X')
    ax.set_ylabel('Y')

    # set ticks and labels for x and y axes
    xticks = np.arange(-x_lim, x_lim + 1, 1)
    yticks = np.arange(-y_lim, y_lim + 1, 1)
    ax.set_xticks(xticks)
    ax.set_yticks(yticks)
    ax.set_xticklabels([str(x) for x in xticks])
    ax.set_yticklabels([str(y) for y in yticks])

    # add gridlines
    ax.grid(True)

    if multi == True:
        for i, y_batch in enumerate(y):
            if label != None:
                plt.plot(x, y_batch, label=label[i])
            else:
                plt.plot(x, y_batch)
        plt.legend()

    else:
        plt.plot(x, y)

    plt.show()
```

```python
# Sigmoid
def sigmoid(x):
  return 1/(1+np.exp(-x))
```

```python
# Draw sigmoid
x = np.arange(-10, 10, 0.01)
y = sigmoid(x)

draw(x, y, 7, 1.5)
```
