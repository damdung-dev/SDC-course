```python
#Exercise 1:
import numpy as np
a = np.array([1, 2])
b = np.array([2, 1])
c = np.array([-2, -1])
dot_ab = np.dot(a,b)
dot_bc = np.dot(b,c)
print(dot_ab)
print(dot_bc)
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)
cos_ab = dot_ab/(norm_a*norm_b)
print(cos_ab)
radian_ab = np.arccos(cos_ab)
degree_ab = np.degrees(radian_ab)
print(f"cos_ab = {cos_ab}")
print(f"radian_ab = {radian_ab}")
print(f"degree_ab = {degree_ab}")
```

```python
# Matrix and Inverse Matrix
# 2*x - y = 0
#   x + y = 3
# ==> x = 1, y = 2
A = np.array([[2,-1], [1,1]])
B = np.array([[0], [3]])
# Ax = b ==> A^-1*A*x = A^-1*b ==> I*x = A^-1*b ==> x = A^-1 * b
inv_A = np.linalg.inv(A)
print(inv_A)
I = np.round(np.dot(inv_A, A), 3)
print(I)

X = np.dot(inv_A, B)
print("Nghiem cua he phuong trinh la :")
print(X)
```

```python
import pandas as pd
AppleStock = [70 , 72 , 80 , 75 , 69 , 85 ]
MSStock    = [120, 125, 130, 128, 110, 150]
GOStock    = [200, 150, 120, 170, 250, 280]

df = pd.DataFrame({"AppleStock" : AppleStock, "MSStock" : MSStock, "GOStock" : GOStock})
display(df)
CovarianceDF = df.cov()
CorrelationDF = df.corr()

display(CovarianceDF)
display(CorrelationDF)
```
