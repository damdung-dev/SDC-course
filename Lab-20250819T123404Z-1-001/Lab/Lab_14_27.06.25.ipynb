```python
import random
import numpy as np
np.random.seed(6)
X = np.random.randint(10, 100, 100)
X = X.reshape(20,5)
print(X)
```

```python
# B1 : De-Mean (Chuẩn hóa dữ liệu)
X_mean = X - np.mean(X, axis = 0)
print(X_mean)
```

```python
# B2 : Calculate Covariance Matrix (Tính ma trận hiệp phương sai)
Cov = np.cov(X_mean.T)
print(Cov.shape)
print(Cov)
```

```python
# B3 : Find Eigen Values and Eigen Vector (Tính vector riêng và giá trị riêng)
eigen_value, eigen_vector = np.linalg.eigh(Cov)
print(eigen_value)
print(eigen_vector)
```

```python
# B4 : Select Best Eigen Values and Eigen Vectors for Dimensionality Reduction
# (Sắp xếp giá trị riêng và vector riêng theo thứ tự ưu tiên giảm dần)
# Bộ vector mới còn gọi là "Feature vectors"
n_selectDim = 3
sorted_index = np.flip(np.argsort(eigen_value)) # 4 3 2 1 0
selected_eigenvectors = eigen_vector.T[sorted_index[:n_selectDim]]
print(selected_eigenvectors)
```

```python
# B5 : Project data onto Selected Eigen Vectors to reduce dimesionality
X_mean_projected = np.dot(selected_eigenvectors, X_mean.T).T
print("The data after projected onto eigen vectors : Dimensionality reduction")
print(X_mean_projected)
```
